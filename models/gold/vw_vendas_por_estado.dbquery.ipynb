```sql
CREATE OR REPLACE VIEW workshop_modelagem_aovivo.gold.vw_receita_por_estado AS
SELECT
  dt.data,
  dc.state,
  SUM(f.revenue_net) AS receita_liquida
FROM workshop_modelagem_aovivo.gold.fact_vendas f
JOIN workshop_modelagem_aovivo.gold.dim_tempo   dt ON dt.tempo_sk    = f.fk_tempo_sk
JOIN workshop_modelagem_aovivo.gold.dim_cliente dc ON dc.customer_sk = f.fk_customer_sk
GROUP BY dt.data, dc.state;
```
